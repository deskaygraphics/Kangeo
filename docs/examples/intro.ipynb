```python
import kangeo
import  ipyleaflet
from ipyleaflet import GeoJSON
import geopandas as gpd
```

```python
m = kangeo.Map(center=(40, -100), zoom=4)
url = "https://github.com/opengeos/datasets/releases/download/book/us_states.geojson"
gdf = gpd.read_file(url)
data = gdf.__geo_interface__
gj = GeoJSON(data=data)
m.add_basemap("OpenTopoMap")
m.add_layer(gj)
m
```

```python
gdf.total_bounds
```

```python
m.fit_bounds(gdf.total_bounds)
```
